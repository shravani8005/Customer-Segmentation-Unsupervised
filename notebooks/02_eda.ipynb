## Key Exploratory Data Analysis Observations

1. **Age Distribution**
   - Customer age follows an approximately normal distribution.
   - Majority of customers fall within the middle-age group, indicating a stable customer base.

2. **Balance and Duration**
   - Balance and call duration show right-skewed distributions.
   - A small group of customers have significantly higher balances and longer call durations, indicating potential high-value segments.

3. **Categorical Feature Behavior**
   - Features such as job, marital status, education, contact type, and month appear as discrete values due to label encoding.
   - These features will help differentiate customer behavioral patterns during clustering.

4. **Campaign Interaction Features**
   - Campaign, pdays, and previous interactions are highly skewed.
   - Most customers were contacted only a few times, suggesting limited repeated marketing exposure.

5. **Binary Features**
   - Default, housing loan, personal loan, and deposit features show clear binary separation.
   - These variables are useful indicators for customer financial stability and response behavior.

6. **Overall Insight**
   - The dataset contains a mix of demographic, financial, and behavioral attributes.
   - Feature scaling and preprocessing have made the data suitable for unsupervised clustering algorithms.
