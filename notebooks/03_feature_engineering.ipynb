```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/bank_marketing_cleaned.csv")
df.head()
```

```python
df.columns
```

```python
df['interaction_score'] = df['campaign'] + df['previous']

df['avg_contact_intensity'] = df['duration'] / (df['campaign'] + 1)
```

```python
df['balance_per_contact'] = df['balance'] / (df['campaign'] + 1)
```

```python
# Lower pdays = more recent contact
df['recency_score'] = -df['pdays']
```

```python
df.describe()
```

```python
df.to_csv("../data/processed/bank_marketing_features.csv", index=False)
```

## Feature Engineering Summary

1. **Interaction Score**
   - Combines campaign and previous contact counts.
   - Represents overall marketing exposure of a customer.

2. **Average Contact Intensity**
   - Measures engagement effectiveness per contact.
   - Helps identify customers who respond better to fewer interactions.

3. **Balance per Contact**
   - Indicates customer financial value relative to marketing effort.

4. **Recency Score**
   - Derived from pdays to represent how recently the customer was contacted.
   - More recent interactions imply higher engagement potential.

These engineered features enhance customer behavior representation and improve clustering quality.
